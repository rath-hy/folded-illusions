# Folding story: a new picture at every stage

One printed square that tells a short story as you fold it into a crane. Each stage of folding shows a different picture:

| Stage | Example |
|---|---|
| Flat sheet | a tangle of leaves |
| Bird base | a chrysalis |
| Crane | a butterfly |

This extends [`illusions.ipynb`](illusions.ipynb), which makes one picture flat and two on the folded crane. Here the folded views come from different models: the crane, and a base you fold on the way to it.

**The catch:** every picture lives on the same sheet, so pictures compete for pixels in two ways.

- **Against the flat sheet.** The flat view shows every pixel, so each stage's picture has to hide in the flat picture. How hard that is depends on how much folding scrambles the stage's visible paper (table below).
- **Between consecutive stages.** Folding keeps much of the outside on the outside, so neighboring stages show a lot of the same paper. With a crane, up to about half of a stage's visible paper also shows at another stage. Those pixels are painted once for both pictures. An early test with the digits 1, 2, 3 made the 2 and the 3 blur into each other for this reason.

| Stage | Pieces of paper on the visible side | Hiding the picture when flat |
|---|---|---|
| Preliminary base | 1 (an uncut quarter of the sheet, turned 135°) | hard: the picture *is* a corner of the flat picture |
| Bird base | 2 | medium |
| Crane | about 10 | easy, as in trial 1 |

**So by default the story has three parts: flat, bird base, crane.** The preliminary base is left out. Its picture can't hide (it sits on the flat sheet intact), and it shares about half its paper with the bird base. Without it, the bird base and the crane share no paper at all, and the flat picture has two thirds of the sheet to itself. To bring it back, add `"preliminary base"` to `STAGE_NAMES` below. What makes a stage good isn't being small, it's scrambling the paper into pieces.

## The stages

The bases' crease patterns and folded states are in `stages/`, made by `make_stages.py` with Flat-Folder's solver (the same tool that gave us the crane). Their creases match the crane's, so they all share one sheet.

Each stage can be held either way round. We pick the sides so that no stage shares too much of its visible paper with another stage. Each stage is shown upright: the preliminary base with its closed point up, the bird base with its wing flaps up, the crane as Flat-Folder draws it.

In the maps below, color is the paper a stage shows, on the printed side; everything else is hidden at that stage.

```python
STAGE_NAMES = DEFAULT_STAGES     # ["bird base", "crane"]; or add "preliminary base" first
stages, models = load_stages(STAGE_NAMES)
SIDES = least_shared_sides(models)
names = [name for name, _, _, _ in stages]

fig, axes = plt.subplots(1, len(models), figsize=(5 * len(models), 5))
seen = []
for ax, m, side, name in zip(axes, models, SIDES, names):
    _, side2, _, _ = m.sheet_visibility(300)
    shown = side2 == (1 if side == "front" else 2)
    ax.imshow(np.where(shown[..., None], PALETTE[1 if side == "front" else 2], 0.95))
    ax.set_title(f"{name}, {side}: {shown.mean():.0%} of the sheet")
    ax.axis("off")
plt.show()
for name, share in zip(names, shared_paper(models, SIDES)):
    print(f"{name}: {share:.0%} of its visible paper also shows at another stage")
```

## 2. Pictures that hide each other (needs a GPU)

This is the sampler from `illusions.ipynb` with more views. At each denoising step, DeepFloyd IF sees the sheet several ways, each with its own prompt: flat, and folded into each stage. Their predictions are mapped back onto the sheet and blended:

- **Pixels hidden at every stage** follow only the flat picture.
- **A pixel shown at some stages** blends the flat prediction with each of those stages' predictions. `FOLD_WEIGHT` sets how much the folded pictures count.

Because every view is generated together, the model can choose content that works in all of them. For example, the chrysalis's green can double as leaves in the flat scene.

The bird base is the harder stage to hide: its picture appears on the flat sheet as two halves, along two edges. The crane's is scattered over about ten pieces.

### The story

One prompt for the flat sheet, then one per stage in `STAGE_NAMES`, in folding order. Tips:

- **Share a style** across all prompts. Painterly styles hide seams.
- **Make the flat scene busy.** Leaves, twigs, a crowd, a mosaic: texture gives the later pictures somewhere to hide.
- **Give each stage a distinct subject and color,** so each fold is a visible change. An earlier run (a bud opening into a flower) came out as the same pink flower at every stage.
- **If stages share paper** (the preliminary base and bird base do), their scenes should be able to share strokes.
- **Prefer flexible subjects** that can be drawn many ways (flowers, animals) over rigid symbols like digits and letters.
- **Fit each subject to its stage's outline.** The bird base is a tall skinny diamond (a chrysalis, a candle flame, a feather), the crane a bird with wings (a butterfly, a phoenix), the preliminary base a square diamond. In the flower run, the bird base came out as a bud on a stem whatever it was asked for, because that's what fits.

```python
STYLE = "a watercolor painting of "
FLAT_PROMPT = STYLE + "a tangle of green leaves and stems"
STAGE_PROMPTS = [
    STYLE + "a green chrysalis hanging from a twig, on a plain light background",          # bird base
    STYLE + "an orange monarch butterfly with open wings, on a plain light background",   # crane
]
# Another story that fits these outlines: firewood logs (flat), a tall candle flame, a phoenix.
assert len(STAGE_PROMPTS) == len(stages), f"one prompt per stage: {names}"
```

### Encode the prompts

As in `illusions.ipynb`: the text encoder is loaded onto the GPU, used once, and deleted to save memory. **Rerun this cell whenever you change the prompts.** The first run downloads about 11 GB.

```python
import gc, torch
from transformers import T5EncoderModel
from diffusers import IFPipeline, IFSuperResolutionPipeline

skip = dict(safety_checker=None, feature_extractor=None, watermarker=None)

text_encoder = T5EncoderModel.from_pretrained("DeepFloyd/IF-I-M-v1.0", subfolder="text_encoder", variant="fp16",
                                              torch_dtype=torch.float16, device_map="cuda")
encoder = IFPipeline.from_pretrained("DeepFloyd/IF-I-M-v1.0", text_encoder=text_encoder, unet=None, **skip)
prompt_embeds, negative_embeds = encoder.encode_prompt([FLAT_PROMPT] + STAGE_PROMPTS)

del encoder, text_encoder
gc.collect(); torch.cuda.empty_cache()
print("embeddings:", tuple(prompt_embeds.shape))
```

### Load the image models

Once per session.

```python
if "stage_1" not in globals():
    stage_1 = IFPipeline.from_pretrained("DeepFloyd/IF-I-M-v1.0", text_encoder=None, variant="fp16",
                                         torch_dtype=torch.float16, **skip).to("cuda")
    stage_2 = IFSuperResolutionPipeline.from_pretrained("DeepFloyd/IF-II-M-v1.0", text_encoder=None, variant="fp16",
                                                        torch_dtype=torch.float16, **skip).to("cuda")
```

### Generate

- `FOLD_WEIGHT`: how much shown pixels follow the stage pictures (1) vs. the flat one (0). Raise it if a stage's picture is swallowed by the flat scene.
- `GUIDANCE`: how strongly each stage follows the prompts (stage 1, stage 2).
- `SEED`: try several. Some seeds hide the stage pictures much better than others.

The previews fold the sheet with the decal simulator, independently of the maps used to make it.

```python
from illusion import generate

FOLD_WEIGHT = 0.5
GUIDANCE = (7.0, 4.0)
STEPS = (50, 40)
SEED = 0

views64, views256 = stage_views(stages, models, SIDES, 64), stage_views(stages, models, SIDES, 256)
sheet64, sheet256 = generate(stage_1, stage_2, views64, views256, (negative_embeds, prompt_embeds),
                             guidance=GUIDANCE, steps=STEPS, fold_weight=FOLD_WEIGHT, seed=SEED)
titles = ["flat"] + [f"{name} ({side})" for name, side in zip(names, SIDES)]
plot_story(sheet64, stages, models, SIDES, titles);
plot_story(sheet256, stages, models, SIDES, titles);
```

### Print it

Saves the sheet, and an A4 PDF with it at 200 mm, centered, with a thin gray cut line. Print at **actual size** (not "fit to page"), cut along the line, and fold with the **printed side out**. A big square folds more precisely: a millimeter of error is a smaller share of each piece.

To check which side of each stage to look at: the bases are held point up; the side shown is the one in the preview above. If a stage looks wrong, turn it over.

```python
name = f"story_seed{SEED}"
img = print_page(sheet256, name + "_a4.pdf")    # 200 mm square at 300 dpi
img.save(name + ".png", dpi=(300, 300))
print("saved", name + ".png", "and", name + "_a4.pdf")
try:
    from google.colab import files
    files.download(name + "_a4.pdf")
except ImportError:
    pass
```

## Notes

- **The sampler treats all views as equals.** Hiding each picture until its stage comes is something we hope the result does, not something it enforces. If a stage's picture shows on the flat sheet too much, lower `FOLD_WEIGHT` or try another seed.
- **Shared paper sets what's possible.** Pixels that two stages both show can only be one color. Check the shared-paper numbers above when choosing stages and scenes.
- **For a fourth beat**, add a stage that scrambles the paper rather than the preliminary base, such as the bird base with its legs narrowed (a flat step between the bird base and the crane). `make_stages.py` makes new stages from crease lines.
- **Measure it.** Generate many seeds and score each view against its prompt with CLIP, as Visual Anagrams does. That separates luck from settings.
